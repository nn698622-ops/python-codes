````markdown
# Python Lists — Comprehensive Guide

A **list** is an ordered, mutable collection in Python. Lists can store elements of different data types and allow duplicate values.

```python
my_list = [10, 20, 30, 40, 50]
mixed_list = [10, "Python", 3.14, True]
````

---

## 1. Creating a List

```python
# Empty list
my_list = []

# List with values
numbers = [10, 20, 30, 40]

# List with different data types
data = [10, "Python", 3.14, True]

# Using the list() constructor
numbers = list((10, 20, 30))
```

---

## 2. Accessing List Elements

List indexing starts from `0`.

```python
numbers = [10, 20, 30, 40, 50]

numbers[0]      # 10
numbers[1]      # 20
numbers[-1]     # 50
numbers[-2]     # 40
```

### Index positions

```text
Value:       10    20    30    40    50
Index:        0     1     2     3     4
Negative:    -5    -4    -3    -2    -1
```

---

## 3. List Slicing

Syntax:

```python
list[start:stop:step]
```

The `stop` index is **not included**.

```python
numbers = [10, 20, 30, 40, 50]

numbers[1:4]     # [20, 30, 40]
numbers[:3]      # [10, 20, 30]
numbers[2:]      # [30, 40, 50]
numbers[:]       # Copy of the list
numbers[::2]     # [10, 30, 50]
numbers[::-1]    # [50, 40, 30, 20, 10]
```

---

# 4. Modifying a List

Lists are **mutable**, meaning their elements can be changed after creation.

### Change an element

```python
numbers = [10, 20, 30, 40]

numbers[1] = 25

# [10, 25, 30, 40]
```

### Change multiple elements

```python
numbers[1:3] = [25, 35]

# [10, 25, 35, 40]
```

---

# 5. Adding Elements

## `append()`

Adds **one element** to the end of the list.

```python
numbers = [10, 20, 30]

numbers.append(40)

# [10, 20, 30, 40]
```

> `append()` adds the entire object as a single element.

```python
numbers.append([50, 60])

# [10, 20, 30, 40, [50, 60]]
```

---

## `insert()`

Adds an element at a specific index.

```python
numbers = [10, 20, 30]

numbers.insert(1, 15)

# [10, 15, 20, 30]
```

Syntax:

```python
list.insert(index, value)
```

---

## `extend()`

Adds multiple elements from another iterable.

```python
numbers = [10, 20, 30]

numbers.extend([40, 50, 60])

# [10, 20, 30, 40, 50, 60]
```

### `append()` vs `extend()`

```python
a = [1, 2, 3]

a.append([4, 5])
# [1, 2, 3, [4, 5]]
```

```python
a = [1, 2, 3]

a.extend([4, 5])
# [1, 2, 3, 4, 5]
```

---

# 6. Removing Elements

## `remove()`

Removes the **first occurrence** of a specified value.

```python
numbers = [10, 20, 30, 20, 40]

numbers.remove(20)

# [10, 30, 20, 40]
```

If the value does not exist, Python raises:

```text
ValueError
```

---

## `pop()`

Removes and returns an element.

```python
numbers = [10, 20, 30, 40]

numbers.pop()
# Returns 40

numbers.pop(1)
# Returns 20
```

After these operations:

```python
# [10, 30]
```

If no index is specified, `pop()` removes the last element.

---

## `del`

Deletes an element or a slice.

```python
numbers = [10, 20, 30, 40]

del numbers[1]

# [10, 30, 40]
```

Delete multiple elements:

```python
del numbers[1:3]
```

Delete the entire list:

```python
del numbers
```

---

## `clear()`

Removes all elements.

```python
numbers = [10, 20, 30]

numbers.clear()

# []
```

---

# 7. Searching in a List

## `in`

Checks whether an element exists.

```python
numbers = [10, 20, 30, 40]

20 in numbers
# True

50 in numbers
# False
```

## `not in`

```python
50 not in numbers
# True
```

---

## `index()`

Returns the index of the first occurrence.

```python
numbers = [10, 20, 30, 20, 40]

numbers.index(20)
# 1
```

You can specify a starting position:

```python
numbers.index(20, 2)
# 3
```

---

## `count()`

Counts how many times a value appears.

```python
numbers = [10, 20, 20, 30, 20]

numbers.count(20)
# 3
```

---

# 8. Sorting Lists

## `sort()`

Sorts the original list.

```python
numbers = [40, 10, 30, 20]

numbers.sort()

# [10, 20, 30, 40]
```

### Descending order

```python
numbers.sort(reverse=True)

# [40, 30, 20, 10]
```

---

## `sorted()`

Returns a **new sorted list** without modifying the original.

```python
numbers = [40, 10, 30, 20]

new_numbers = sorted(numbers)

# numbers     → [40, 10, 30, 20]
# new_numbers → [10, 20, 30, 40]
```

### Important difference

| `sort()`               | `sorted()`          |
| ---------------------- | ------------------- |
| Modifies original list | Creates a new list  |
| Returns `None`         | Returns sorted list |
| List method            | Built-in function   |

---

# 9. Reversing a List

## `reverse()`

Reverses the original list.

```python
numbers = [10, 20, 30, 40]

numbers.reverse()

# [40, 30, 20, 10]
```

## Using slicing

```python
reversed_list = numbers[::-1]
```

## Using `reversed()`

```python
reversed_list = list(reversed(numbers))
```

# 14. List Comprehension

List comprehension provides a concise way to create lists.

### Basic syntax

```python
[expression for item in iterable]
```

Example:

```python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers]

# [1, 4, 9, 16, 25]
```

---

## List Comprehension with Condition

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in numbers if x % 2 == 0]

# [2, 4, 6]
```

### With `if-else`

```python
result = ["Even" if x % 2 == 0 else "Odd" for x in numbers]

# ['Odd', 'Even', 'Odd', 'Even', 'Odd', 'Even']
```

Access elements:

```python
matrix[0]
```

---

## Iterating Through a Nested List

```python
for row in matrix:
    for value in row:
        print(value)
```

# 16. Copying Lists

## Assignment does NOT create an independent copy

```python
a = [1, 2, 3]

b = a

b.append(4)

print(a)
# [1, 2, 3, 4]
```

Both variables refer to the same list.

---

## Using `copy()`

```python
a = [1, 2, 3]

b = a.copy()

b.append(4)

print(a)
# [1, 2, 3]

print(b)
# [1, 2, 3, 4]
```

---

## Using slicing

```python
b = a[:]
```

## Using `list()`

```python
b = list(a)
```

---

# 17. Shallow Copy vs Deep Copy

For nested lists, a shallow copy does not fully duplicate nested objects.

```python
a = [[1, 2], [3, 4]]

b = a.copy()

b[0][0] = 100

print(a)
# [[100, 2], [3, 4]]
```

Use `deepcopy()` when an independent nested copy is required.

```python
import copy

b = copy.deepcopy(a)
```

# 19. Sorting with `key`

You can control how elements are sorted.

```python
words = ["apple", "kiwi", "banana", "fig"]

words.sort(key=len)

# ['fig', 'kiwi', 'apple', 'banana']
```

### Sorting strings ignoring case

```python
names = ["john", "Alice", "bob"]

names.sort(key=str.lower)
```

| ------------------ | ------------------------------------- |
| `append(x)`        | Adds `x` to the end                   |

# 22. Common Mistakes

### Mistake 1: Using an invalid index

```python
numbers = [10, 20, 30]

numbers[5]
```

Raises:

```text
IndexError
```

---

### Mistake 2: Confusing `remove()` and `pop()`

```python
numbers.remove(20)  # Removes the value 20
numbers.pop(1)      # Removes the element at index 1
```

---

### Mistake 3: Confusing `append()` and `extend()`

```python
numbers.append([4, 5])
# Adds [4, 5] as one element

numbers.extend([4, 5])
# Adds 4 and 5 separately
```

---

### Mistake 4: Expecting `sort()` to return a list

```python
numbers = [3, 1, 2]

result = numbers.sort()

print(result)
# None
```

`sort()` modifies the list in place.

Use:

```python
result = sorted(numbers)
```

if you need the returned sorted list.

* **Lists are ordered** collections.
* **Lists are mutable**, so their contents can be changed.
* Lists **allow duplicate values**.
* Indexing starts from **0**.
* Negative indexing starts from **-1**.
* `append()` adds one item.
* `extend()` adds multiple items.
* `insert()` adds an item at a specific position.
* `remove()` removes by value.
* `pop()` removes by index and returns the removed item.
* `sort()` modifies the original list.
* `sorted()` creates a new sorted list.
* `reverse()` reverses the original list.
* List comprehensions provide a concise way to create lists.
* Use `copy()` or other copying techniques when you need an independent list.
