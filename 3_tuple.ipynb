````markdown
# Python Tuples — Comprehensive Guide

A **tuple** is an ordered and immutable collection in Python.

Tuples are similar to lists, but unlike lists, their elements **cannot be changed, added, or removed** after creation.

```python
my_tuple = (10, 20, 30, 40, 50)
````

---

## 1. Creating a Tuple

```python
# Empty tuple
empty_tuple = ()

# Tuple with values
numbers = (10, 20, 30, 40)

# Tuple with different data types
data = (10, "Python", 3.14, True)

# Tuple without parentheses
numbers = 10, 20, 30

# Using the tuple() constructor
numbers = tuple([10, 20, 30])
```

### Single-element Tuple

A comma is required to create a single-element tuple.

```python
a = (10,)      # Tuple
b = (10)       # Integer
```

```python
type(a)        # <class 'tuple'>
type(b)        # <class 'int'>
```

---

# 2. Accessing Tuple Elements

Tuple indexing starts from `0`.

```python
numbers = (10, 20, 30, 40, 50)

numbers[0]      # 10
numbers[1]      # 20
numbers[-1]     # 50
numbers[-2]     # 40
```

### Index positions

```text
Value:       10    20    30    40    50
Index:        0     1     2     3     4
Negative:    -5    -4    -3    -2    -1
```

---

# 3. Tuple Slicing

Syntax:

```python
tuple[start:stop:step]
```

The `stop` index is not included.

```python
numbers = (10, 20, 30, 40, 50)

numbers[1:4]     # (20, 30, 40)
numbers[:3]      # (10, 20, 30)
numbers[2:]      # (30, 40, 50)
numbers[:]       # (10, 20, 30, 40, 50)
numbers[::2]     # (10, 30, 50)
numbers[::-1]    # (50, 40, 30, 20, 10)
```

---

# 4. Tuples Are Immutable

Once a tuple is created, its elements cannot be changed.

```python
numbers = (10, 20, 30)

numbers[0] = 100
```

This raises:

```text
TypeError
```

You cannot directly:

```python
numbers.append(40)
numbers.remove(20)
numbers.pop()
```

because tuples do not support these operations.

---

# 5. Adding Elements to a Tuple

You cannot directly add an element to a tuple.

However, you can create a **new tuple** using concatenation.

```python
numbers = (10, 20, 30)

numbers = numbers + (40,)

# (10, 20, 30, 40)
```

Add multiple elements:

```python
numbers = numbers + (50, 60)

# (10, 20, 30, 40, 50, 60)
```

> The original tuple is not modified. A new tuple is created.

---

# 6. Tuple Concatenation

Use the `+` operator to combine tuples.

```python
a = (1, 2, 3)
b = (4, 5, 6)

c = a + b

# (1, 2, 3, 4, 5, 6)
```

Both operands must be tuples.

```python
(1, 2) + (3, 4)     # Works

(1, 2) + [3, 4]     # TypeError
```

---

# 7. Tuple Repetition

Use `*` to repeat a tuple.

```python
numbers = (1, 2, 3)

numbers * 3

# (1, 2, 3, 1, 2, 3, 1, 2, 3)
```

---

# 8. Searching in a Tuple

## `in`

Checks whether an element exists.

```python
numbers = (10, 20, 30, 40)

20 in numbers
# True

50 in numbers
# False
```

## `not in`

```python
50 not in numbers
# True
```

---

# 9. `index()`

Returns the index of the first occurrence of a value.

```python
numbers = (10, 20, 30, 20, 40)

numbers.index(20)

# 1
```

You can specify where the search should start.

```python
numbers.index(20, 2)

# 3
```

---

# 10. `count()`

Returns the number of times a value occurs.

```python
numbers = (10, 20, 20, 30, 20)

numbers.count(20)

# 3
```

---

# 11. Useful Built-in Functions

Tuples work with many Python built-in functions.

```python
numbers = (10, 20, 30, 40, 50)

len(numbers)     # 5
min(numbers)     # 10
max(numbers)     # 50
sum(numbers)     # 150
```

### `sorted()`

`sorted()` returns a **list**, not a tuple.

```python
numbers = (40, 10, 30, 20)

sorted(numbers)

# [10, 20, 30, 40]
```

If you want a tuple:

```python
tuple(sorted(numbers))

# (10, 20, 30, 40)
```

---

# 12. Iterating Through a Tuple

```python
numbers = (10, 20, 30, 40)

for number in numbers:
    print(number)
```

Output:

```text
10
20
30
40
```

---

## Using `enumerate()`

```python
for index, value in enumerate(numbers):
    print(index, value)
```

Output:

```text
0 10

```

# 13. Tuple Unpacking

Tuple unpacking allows you to assign tuple elements to variables.

```python
numbers = (10, 20, 30)

a, b, c = numbers

print(a)    # 10
print(b)    # 20
print(c)    # 30
```

The number of variables must normally match the number of elements.

```python
a, b = (10, 20, 30)
```

This raises:

```text
ValueError
```

---

# 14. Extended Tuple Unpacking

Use `*` to collect multiple values.

```python
numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers

print(first)     # 1
print(middle)    # [2, 3, 4]
print(last)      # 5
```

> Notice that `middle` is a **list**, even though `numbers` is a tuple.

---

# 15. Swapping Variables Using Tuples

Tuples make swapping values easy.

```python
a = 10
b = 20

a, b = b, a

print(a)    # 20
print(b)    # 10
```

This is a common Python technique.

---

# 16. Nested Tuples

A tuple can contain other tuples.

```python
data = (
    (1, 2),
    (3, 4),
    (5, 6)
)
```

Access nested elements:

```python
data[0]
# (1, 2)

data[0][1]
# 2

data[2][0]
# 5
```

---

# 17. Tuple Containing a List

Although tuples are immutable, they can contain mutable objects such as lists.

```python
data = (10, 20, [30, 40])
```

You cannot replace the list itself:

```python
data[2] = [50, 60]
```

This raises:

```text
TypeError
```

But you **can modify the list inside the tuple**:

```python
data[2].append(50)

print(data)

# (10, 20, [30, 40, 50])
```

> The tuple itself remains unchanged; the mutable list inside it is modified.

---

# 18. Converting Between List and Tuple

## List → Tuple

```python
numbers = [10, 20, 30]

numbers_tuple = tuple(numbers)

# (10, 20, 30)
```

## Tuple → List

```python
numbers = (10, 20, 30)

numbers_list = list(numbers)

# [10, 20, 30]
```

This is useful when you need to modify data temporarily.

```python
numbers = (10, 20, 30)

numbers = list(numbers)
numbers.append(40)
numbers = tuple(numbers)

# (10, 20, 30, 40)
```

# 19. Tuple Membership

You can use `in` and `not in`.

```python
fruits = ("apple", "banana", "orange")

"apple" in fruits
# True

"mango" in fruits
# False

"mango" not in fruits
# True
```

# 20. Comparing Tuples

Tuples can be compared using comparison operators.

```python
a = (1, 2, 3)
b = (1, 2, 4)

a == b      # False
a != b      # True
a < b       # True
```

Python compares tuples **element by element**.

```python
(1, 2, 3) < (1, 2, 4)
# True
```

because `3 < 4`.

---

# 21. Tuple Methods

Tuples have only **two built-in methods**.

| Method     | Purpose                                          |
| ---------- | ------------------------------------------------ |
| `count(x)` | Counts occurrences of `x`                        |
| `index(x)` | Returns the index of the first occurrence of `x` |

Example:

```python
numbers = (10, 20, 20, 30, 20)

numbers.count(20)
# 3

numbers.index(20)
# 1
```

> Unlike lists, tuples do not have methods such as `append()`, `remove()`, `pop()`, `sort()`, or `reverse()`.

# 22. Tuple Packing

Creating a tuple without explicitly using parentheses is called **tuple packing**.

```python
numbers = 10, 20, 30

print(numbers)

# (10, 20, 30)
```

Another example:

```python
person = "John", 25, "India"
```

---

# 23. Tuple Unpacking

The reverse operation is called **tuple unpacking**.

```python
person = ("John", 25, "India")

name, age, country = person

print(name)       # John
print(age)        # 25
print(country)    # India
```

---

# 24. Tuples as Function Return Values

Python functions can return multiple values using a tuple.

```python
def get_user():
    return "John", 25

name, age = get_user()

print(name)    # John
print(age)     # 25
```

The function actually returns:

```python
("John", 25)
```

---

# 25. Tuples as Dictionary Keys

Tuples containing only **hashable elements** can be used as dictionary keys.

```python
locations = {
    (10, 20): "Point A",
    (30, 40): "Point B"
}

print(locations[(10, 20)])

# Point A
```

A list cannot be used as a dictionary key:

```python
locations = {
    [10, 20]: "Point A"
}
```

This raises:

```text
TypeError: unhashable type: 'list'
```

---

# 26. Memory and Performance

Tuples are generally more memory-efficient than lists because they are immutable.

```python
numbers_list = [1, 2, 3, 4, 5]
numbers_tuple = (1, 2, 3, 4, 5)
```

For data that should not change, tuples can be a good choice.

> The performance difference is usually small, so choose based primarily on whether the data should be mutable.

# 27. List vs Tuple

| Feature        | List                 | Tuple            |
| -------------- | -------------------- | ---------------- |
| Syntax         | `[1, 2, 3]`          | `(1, 2, 3)`      |
| Ordered        | Yes                  | Yes              |
| Mutable        | Yes                  | **No**           |
| Duplicates     | Allowed              | Allowed          |
| Indexing       | Yes                  | Yes              |
| Slicing        | Yes                  | Yes              |
| `append()`     | Yes                  | No               |
| `remove()`     | Yes                  | No               |
| `pop()`        | Yes                  | No               |
| `sort()`       | Yes                  | No               |
| `reverse()`    | Yes                  | No               |
| `count()`      | Yes                  | Yes              |
| `index()`      | Yes                  | Yes              |
| Dictionary key | No                   | Yes, if hashable |
| Typical use    | Data that may change | Fixed data       |

# 28. When Should You Use a Tuple?

Use a tuple when:

* The data should **not change**.
* You want to represent a fixed collection of values.
* You want to return multiple values from a function.
* You need a hashable collection to use as a dictionary key.
* You want to communicate that the data is intended to be immutable.

Example:

```python
coordinates = (10.5, 20.3)
rgb_color = (255, 128, 0)
date = (2026, 8, 10)
```

# 29. Common Mistakes

### Mistake 1: Forgetting the comma in a single-element tuple

```python
a = (10)
```

This is an integer.

Correct:

```python
a = (10,)
```

---

### Mistake 2: Trying to modify a tuple

```python
numbers = (10, 20, 30)

numbers[0] = 100
```

Raises:

```text
TypeError
```

---

### Mistake 3: Trying to use list methods

```python
numbers.append(40)
```

Raises:

```text
AttributeError
```

because tuples do not have `append()`.

---

### Mistake 4: Confusing `sorted()` with tuple sorting

```python
numbers = (30, 10, 20)

result = sorted(numbers)

print(result)

# [10, 20, 30]
```

`sorted()` returns a **list**.

To get a tuple:

```python
result = tuple(sorted(numbers))
```

# Key Takeaways

* A **tuple is ordered and immutable**.
* Tuple indexing starts at `0`.
* Negative indexing starts at `-1`.
* Tuples support **indexing and slicing**.
* Tuples allow **duplicate values**.
* Tuples have only two methods: `count()` and `index()`.
* You cannot directly `append()`, `remove()`, `pop()`, or `sort()` a tuple.
* Use `+` to create a new tuple with additional elements.
* Use `tuple()` and `list()` to convert between tuples and lists.
* Tuple unpacking makes it easy to assign multiple values.
* Tuples can be used as dictionary keys when all their elements are hashable.
* A single-element tuple requires a **trailing comma**: `(10,)`.
