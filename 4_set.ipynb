> **Important:** Sets are unordered, so you should not rely on their element positions.

---

# 1. Creating a Set

## Basic Set

```python
numbers = {10, 20, 30, 40}
```

## Set with Different Data Types

```python
data = {10, "Python", 3.14, True}
```

## Empty Set

An empty set must be created using `set()`.

```python
empty_set = set()
```

> `{}` creates an empty **dictionary**, not an empty set.

```python
type({})
# <class 'dict'>

type(set())
# <class 'set'>
```

---

# 2. Creating a Set Using `set()`

You can create a set from another iterable.

```python
numbers = set([10, 20, 30, 40])
```

From a tuple:

```python
numbers = set((10, 20, 30))
```

From a string:

```python
letters = set("hello")

# {'h', 'e', 'l', 'o'}
```

Notice that duplicate characters are automatically removed.

---

# 3. Sets Automatically Remove Duplicates

One of the most important properties of a set is that it stores only **unique values**.

```python
numbers = {10, 20, 20, 30, 30, 30}

print(numbers)

# {10, 20, 30}
```

This makes sets useful for removing duplicates from a list.

```python
numbers = [10, 20, 20, 30, 30, 40]

unique_numbers = set(numbers)

# {10, 20, 30, 40}
```

Convert it back to a list:

```python
unique_numbers = list(set(numbers))
```

> Note: Converting to a set does not preserve the original ordering.

---

# 4. Set Elements Must Be Hashable

Set elements must be **hashable**.

Valid:

```python
numbers = {1, 2, 3}
words = {"apple", "banana"}
data = {(1, 2), (3, 4)}
```

Lists cannot be set elements:

```python
my_set = {[1, 2], [3, 4]}
```

This raises:

```text
TypeError: unhashable type: 'list'
```

Tuples can be elements if their contents are hashable:

```python
my_set = {(1, 2), (3, 4)}
```

---

# 5. Accessing Set Elements

Sets do **not support indexing**.

```python
numbers = {10, 20, 30}

numbers[0]
```

This raises:

```text
TypeError
```

You cannot use:

```python
numbers[0]
numbers[-1]
```

Instead, iterate through the set:

```python
for number in numbers:
    print(number)
```

---

# 6. Checking Membership

Use `in` to check whether an element exists.

```python
numbers = {10, 20, 30, 40}

20 in numbers
# True

50 in numbers
# False
```

Use `not in`:

```python
50 not in numbers
# True
```

Membership checking is one of the most common uses of sets.

---

# 7. Adding Elements

## `add()`

Adds a single element to a set.

```python
numbers = {10, 20, 30}

numbers.add(40)

# {10, 20, 30, 40}
```

If the element already exists, nothing happens.

```python
numbers.add(20)

# Still {10, 20, 30, 40}
```

---

## `update()`

Adds multiple elements from an iterable.

```python
numbers = {10, 20, 30}

numbers.update([40, 50, 60])

# {10, 20, 30, 40, 50, 60}
```

You can also use another set:

```python
numbers.update({70, 80})
```

Or a tuple:

```python
numbers.update((90, 100))
```

---

# 8. Removing Elements

## `remove()`

Removes a specified element.

```python
numbers = {10, 20, 30, 40}

numbers.remove(30)

# {10, 20, 40}
```

If the element does not exist, `remove()` raises:

```text
KeyError
```

---

## `discard()`

Removes an element if it exists.

```python
numbers = {10, 20, 30, 40}

numbers.discard(30)
```

Unlike `remove()`, `discard()` does not raise an error if the element is missing.

```python
numbers.discard(100)
# No error
```

### `remove()` vs `discard()`

| Method       | Element exists | Element doesn't exist |
| ------------ | -------------- | --------------------- |
| `remove(x)`  | Removes it     | Raises `KeyError`     |
| `discard(x)` | Removes it     | Does nothing          |

---

## `pop()`

Removes and returns an arbitrary element.

```python
numbers = {10, 20, 30, 40}

item = numbers.pop()
```

> Because sets are unordered, you should **not assume which element `pop()` will remove**.

---

## `clear()`

Removes all elements.

```python
numbers = {10, 20, 30}

numbers.clear()

# set()
```

---

# 9. Set Union

Union combines all unique elements from two or more sets.

```python
A = {1, 2, 3}
B = {3, 4, 5}

A | B

# {1, 2, 3, 4, 5}
```

Using the method:

```python
A.union(B)
```

Both produce the same result.

---

# 10. Set Intersection

Intersection returns elements that are present in **both sets**.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A & B

# {2, 3}
```

Using the method:

```python
A.intersection(B)
```

---

# 11. Set Difference

Difference returns elements that are in the first set but **not in the second set**.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A - B

# {1}
```

Reverse the order:

```python
B - A

# {4}
```

Using the method:

```python
A.difference(B)
```

> `A - B` is not necessarily the same as `B - A`.

---

# 12. Symmetric Difference

Returns elements that are in either set, but **not in both**.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A ^ B

# {1, 4}
```

Using the method:

```python
A.symmetric_difference(B)
```

---

# 13. Set Operations Summary

Suppose:

```python
A = {1, 2, 3}
B = {3, 4, 5}
```

| Operation            | Operator | Method                      | Result            |
| -------------------- | -------- | --------------------------- | ----------------- |
| Union                | `A \| B` | `A.union(B)`                | `{1, 2, 3, 4, 5}` |
| Intersection         | `A & B`  | `A.intersection(B)`         | `{3}`             |
| Difference           | `A - B`  | `A.difference(B)`           | `{1, 2}`          |
| Difference           | `B - A`  | `B.difference(A)`           | `{4, 5}`          |
| Symmetric Difference | `A ^ B`  | `A.symmetric_difference(B)` | `{1, 2, 4, 5}`    |

# 14. Subset

A set `A` is a subset of `B` if every element of `A` is also present in `B`.

```python
A = {1, 2}
B = {1, 2, 3, 4}

A.issubset(B)

# True
```

Using the operator:

```python
A <= B
# True
```

---

# 15. Proper Subset

A proper subset means `A` is a subset of `B`, but `A` and `B` are not equal.

Use `<`:

```python
A = {1, 2}
B = {1, 2, 3}

A < B
# True
```

---

# 16. Superset

A set `A` is a superset of `B` if `A` contains every element of `B`.

```python
A = {1, 2, 3, 4}
B = {1, 2}

A.issuperset(B)

# True
```

Using the operator:

```python
A >= B
# True
```

---

# 17. Proper Superset

```python
A = {1, 2, 3}
B = {1, 2}

A > B

# True
```

---

# 18. Disjoint Sets

Two sets are **disjoint** if they have no elements in common.

```python
A = {1, 2, 3}
B = {4, 5, 6}

A.isdisjoint(B)

# True
```

If they have at least one common element:

```python
A = {1, 2, 3}
B = {3, 4, 5}

A.isdisjoint(B)

# False
```

---

# 19. Comparing Sets

Sets support equality and inequality comparisons.

```python
A = {1, 2, 3}
B = {3, 2, 1}

A == B
# True
```

Order does not matter.

```python
A != B
# False
```

Sets can also be compared using subset/superset operators:

```python
A <= B
A < B
A >= B
A > B
```

---

# 20. Modifying Sets In-Place

Some set methods modify the original set.

## `update()`

```python
A = {1, 2}

A.update({3, 4})

# {1, 2, 3, 4}
```

## `intersection_update()`

Keeps only elements common to both sets.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A.intersection_update(B)

# {2, 3}
```

## `difference_update()`

Removes elements found in another set.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A.difference_update(B)

# {1}
```

## `symmetric_difference_update()`

Keeps elements that occur in exactly one of the sets.

```python
A = {1, 2, 3}
B = {2, 3, 4}

A.symmetric_difference_update(B)

# {1, 4}
```

# 21. Set Comprehension

Set comprehensions provide a concise way to create sets.

### Syntax

```python
{expression for item in iterable}
```

Example:

```python
numbers = {1, 2, 3, 4, 5}

squares = {x ** 2 for x in numbers}

# {1, 4, 9, 16, 25}
```

---

## Set Comprehension with Condition

```python
numbers = range(1, 11)

even_numbers = {x for x in numbers if x % 2 == 0}

# {2, 4, 6, 8, 10}
```

# 24. Frozenset

A `frozenset` is an **immutable set**.

```python
numbers = frozenset([1, 2, 3, 4])
```

You cannot modify it:

```python
numbers.add(5)
```

This raises:

```text
AttributeError
```

A frozenset supports set operations such as:

```python
A = frozenset([1, 2, 3])
B = frozenset([2, 3, 4])

A & B
# frozenset({2, 3})
```

Because it is immutable and hashable, a frozenset can be used as a dictionary key or as an element of another set.

# 27. List vs Tuple vs Set

| Feature                 | List                     | Tuple      | Set                          |
| ----------------------- | ------------------------ | ---------- | ---------------------------- |
| Ordered                 | Yes                      | Yes        | No                           |
| Mutable                 | Yes                      | No         | Yes                          |
| Duplicates              | Allowed                  | Allowed    | **Not allowed**              |
| Indexing                | Yes                      | Yes        | No                           |
| Slicing                 | Yes                      | Yes        | No                           |
| `append()`              | Yes                      | No         | No                           |
| `add()`                 | No                       | No         | Yes                          |
| Mathematical operations | No                       | No         | **Yes**                      |
| Hashable                | No                       | Yes*       | No                           |
| Main use                | Ordered, changeable data | Fixed data | Unique data / set operations |

`*` A tuple is hashable only when all of its elements are hashable.

# 28. Common Mistakes

## Mistake 1: Creating an empty set incorrectly

```python
my_set = {}
```

This creates a dictionary.

Correct:

```python
my_set = set()
```

---

## Mistake 2: Trying to index a set

```python
numbers = {10, 20, 30}

numbers[0]
```

Sets do not support indexing.

---

## Mistake 3: Assuming set order

```python
numbers = {30, 10, 20}

print(numbers)
```

Do not assume that the output will follow insertion order.

---

## Mistake 4: Using `remove()` when the item may not exist

```python
numbers.remove(100)
```

This can raise `KeyError`.

Use:

```python
numbers.discard(100)
```

if you don't want an error.

---

## Mistake 5: Using a list as a set element

```python
my_set = {[1, 2], [3, 4]}
```

Lists are mutable and therefore unhashable.

Use tuples instead:

```python
my_set = {(1, 2), (3, 4)}
```

# 29. Practical Examples

## Remove Duplicates from a List

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique = set(numbers)

print(unique)

# {1, 2, 3, 4, 5}
```

---

## Find Common Students

```python
python_students = {"Alice", "Bob", "Charlie"}
java_students = {"Bob", "Charlie", "David"}

common = python_students & java_students

# {'Bob', 'Charlie'}
```

---

## Find Students Taking Only Python

```python
python_only = python_students - java_students

# {'Alice'}
```

---

## Find Students Taking Either Course but Not Both

```python
only_one_course = python_students ^ java_students

# {'Alice', 'David'}
```

# 30. Key Takeaways

* A **set stores unique elements**.
* Sets are **mutable**.
* Sets are **unordered collections**.
* Sets do **not support indexing or slicing**.
* Use `add()` to add one element.
* Use `update()` to add multiple elements.
* Use `remove()` or `discard()` to remove elements.
* Use `pop()` to remove an arbitrary element.
* `union` → combines elements.
* `intersection` → finds common elements.
* `difference` → finds elements unique to one set.
* `symmetric difference` → finds elements in either set but not both.
* Sets are excellent for **duplicate removal and membership testing**.
* Use `frozenset` when you need an **immutable set**.
