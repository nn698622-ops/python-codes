````markdown
# Python Dictionaries — Comprehensive Guide

A **dictionary** is a mutable collection of **key-value pairs** in Python.

Dictionaries are useful when you want to store and retrieve data using a meaningful **key** instead of a numeric index.

```python
student = {
    "name": "Alice",
    "age": 20,
    "course": "Python"
}
````

### Basic Structure

```text
key → value

"name"   → "Alice"
"age"    → 20
"course" → "Python"
```

> **Important:** Dictionary keys must be unique and hashable. Values can be of almost any data type.

# 2. Dictionary Keys

Dictionary keys must be **unique**.

```python
student = {
    "name": "Alice",
    "age": 20,
    "age": 25
}

print(student)

# {'name': 'Alice', 'age': 25}
```

If a key is repeated, the **last value replaces the previous value**.

---

## Valid Dictionary Keys

Immutable/hashable objects can generally be used as keys.

```python
data = {
    "name": "Alice",
    101: "Student ID",
    (10, 20): "Coordinates"
}
```

Lists cannot be keys:

```python
data = {
    [1, 2]: "value"
}
```

This raises:

```text
TypeError: unhashable type: 'list'
```

---

# 3. Accessing Dictionary Values

## Using Square Brackets

```python
student = {
    "name": "Alice",
    "age": 20
}

student["name"]
# 'Alice'

student["age"]
# 20
```

If the key does not exist:

```python
student["city"]
```

This raises:

```text
KeyError
```

---

## Using `get()`

`get()` safely retrieves a value.

```python
student.get("name")
# 'Alice'
```

If the key doesn't exist:

```python
student.get("city")
# None
```

You can provide a default value:

```python
student.get("city", "Unknown")

# 'Unknown'
```

### `[]` vs `get()`

| Syntax                           | Missing Key         |
| -------------------------------- | ------------------- |
| `student["city"]`                | Raises `KeyError`   |
| `student.get("city")`            | Returns `None`      |
| `student.get("city", "Unknown")` | Returns `"Unknown"` |

```python
student = {
    "name": "Alice",
    "age": 20
}
```

Result:

```python

    "name": "Alice",
```
```python
    "age": 21,
```

## Using Keyword Arguments

# 7. Removing Dictionary Items

## `pop()`

Removes a specified key and returns its value.

```python
student = {
    "name": "Alice",
    "age": 20,
    "city": "Delhi"
}

age = student.pop("age")

print(age)
# 20
```

Result:

```python
{
    "name": "Alice",
    "city": "Delhi"
}
```

If the key doesn't exist, `pop()` raises `KeyError`.

You can provide a default:

```python
student.pop("country", "Not Found")
```

---

# 8. `popitem()`

Removes and returns the **last inserted key-value pair**.

```python
student = {
    "name": "Alice",
    "age": 20,
    "city": "Delhi"
}

item = student.popitem()

print(item)
# ('city', 'Delhi')
```

Afterward:

```python
{
    "name": "Alice",
    "age": 20
}
```

> Dictionaries preserve insertion order in Python 3.7+.

---

# 9. `del`

Deletes a dictionary item using its key.

```python
student = {
    "name": "Alice",
    "age": 20
}

del student["age"]
```

Result:

```python
{
    "name": "Alice"
}
```

You can also delete the entire dictionary:

```python
del student
```

# 10. `clear()`


# 17. Dictionary Copying

## `copy()`

Creates a shallow copy.

```python
student = {
    "name": "Alice",
    "age": 20
}

new_student = student.copy()
```

Changing the copied dictionary:

```python
new_student["age"] = 25

print(student)
# {'name': 'Alice', 'age': 20}

print(new_student)
# {'name': 'Alice', 'age': 25}
```

---

## Assignment Does Not Create a Copy

```python
student = {
    "name": "Alice"
}

new_student = student

new_student["name"] = "Bob"

print(student)

# {'name': 'Bob'}
```

Both variables refer to the same dictionary.

# 24. `setdefault()`

`setdefault()` returns the value of a key if it exists.

If the key does not exist, it creates the key with a default value.

```python
student = {
    "name": "Alice"
}

student.setdefault("age", 20)

print(student)

# {'name': 'Alice', 'age': 20}
```

If the key already exists:

```python
student.setdefault("name", "Bob")

print(student["name"])

# Alice
```

The existing value is not replaced.

# 26. Merging Dictionaries

Python provides several ways to combine dictionaries.

## Using `update()`

```python
a = {"name": "Alice", "age": 20}
b = {"city": "Delhi", "course": "Python"}

a.update(b)
```

---

## Using `|`

Python 3.9+ supports the dictionary union operator.

```python
a = {"name": "Alice"}
b = {"age": 20}

c = a | b

# {'name': 'Alice', 'age': 20}
```

The original dictionaries are unchanged.

---

## Using `|=`

Updates the dictionary in place.

```python
a = {"name": "Alice"}
b = {"age": 20}

a |= b

# {'name': 'Alice', 'age': 20}
```

---

# 27. Duplicate Keys When Merging

If two dictionaries contain the same key, the value from the **right-hand dictionary wins**.

```python
a = {
    "name": "Alice",
    "age": 20
}

b = {
    "age": 25,
    "city": "Delhi"
}

c = a | b
```

Result:

```python
{
    "name": "Alice",
    "age": 25,
    "city": "Delhi"
}
```

# 28. Dictionary Sorting

Dictionaries can be sorted using `sorted()`.

### Sort by Keys

```python
student = {
    "city": "Delhi",
    "age": 20,
    "name": "Alice"
}

sorted(student)
```

Result:

```python
['age', 'city', 'name']
```

---

## Sort Key-Value Pairs by Key

```python
sorted(student.items())
```

---

## Sort by Values

```python
marks = {
    "Alice": 90,
    "Bob": 75,
    "Charlie": 85
}

sorted(
    marks.items(),
    key=lambda item: item[1]
)
```

Result:

```python
[
    ('Bob', 75),
    ('Charlie', 85),
    ('Alice', 90)
]
```

Descending order:

```python
sorted(
    marks.items(),
    key=lambda item: item[1],
    reverse=True
)
```

# 34. Dictionary Operators

```python
student = {
    "name": "Alice",
    "age": 20
}
```

### Membership

```python
"name" in student
# True

"city" not in student
# True
```

### Dictionary Union

```python
a = {"x": 1}
b = {"y": 2}

a | b
# {'x': 1, 'y': 2}
```

### Equality

```python
{"a": 1, "b": 2} == {"b": 2, "a": 1}

# True
```

Dictionary equality does not depend on insertion order.

# 35. Dictionary vs List vs Tuple vs Set

| Feature    | List         | Tuple      | Set           | Dictionary      |
| ---------- | ------------ | ---------- | ------------- | --------------- |
| Ordered    | Yes          | Yes        | No*           | Yes             |
| Mutable    | Yes          | No         | Yes           | Yes             |
| Duplicates | Yes          | Yes        | No            | Keys: No        |
| Indexing   | Yes          | Yes        | No            | By key          |
| Stores     | Values       | Values     | Unique values | Key-value pairs |
| Main use   | Ordered data | Fixed data | Unique data   | Mapped data     |

`*` Sets are unordered; do not rely on iteration order.

# 36. Common Mistakes

## Mistake 1: Accessing a missing key with `[]`

```python
student["city"]
```

If `"city"` doesn't exist, this raises `KeyError`.

Use:

```python
student.get("city")
```

---

## Mistake 2: Using a list as a key

```python
data = {
    [1, 2]: "value"
}
```

Raises:

```text
TypeError: unhashable type: 'list'
```

Use a tuple:

```python
data = {
    (1, 2): "value"
}
```

---

## Mistake 3: Confusing keys and values

```python
student = {
    "name": "Alice",
    "age": 20
}

"name" in student
# True
```

But:

```python
"Alice" in student
# False
```

To search values:

```python
"Alice" in student.values()
# True
```

---

## Mistake 4: Modifying a dictionary while iterating over it

Avoid:

```python
for key in student:
    del student[key]
```

This can raise:

```text
RuntimeError: dictionary changed size during iteration
```

Instead, iterate over a copy of the keys:

```python
for key in list(student):
    del student[key]
```

# Key Takeaways

* A **dictionary stores data as key-value pairs**.
* Dictionary **keys must be unique**.
* Dictionary keys must be **hashable**.
* Dictionary values can be duplicated and can contain almost any data type.
* Dictionaries are **mutable**.
* Use `dict[key]` when you expect the key to exist.
* Use `get()` when the key may not exist.
* Use `update()` to add or modify multiple key-value pairs.
* Use `keys()`, `values()`, and `items()` to access dictionary contents.
* Use `pop()` or `popitem()` to remove elements.
* Use `in` to check whether a **key** exists.
* Use dictionary comprehensions to create dictionaries concisely.
* Dictionaries can contain nested dictionaries and other data structures.
* Python 3.7+ dictionaries preserve **insertion order**.
* Use dictionaries when you need to associate a **key with a value**.
