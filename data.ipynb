## Difference Between List, Tuple, Set, and Dictionary in Python

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| **Syntax** | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` | `{"a": 1, "b": 2}` |
| **Ordered** | Yes | Yes | No | Yes* |
| **Mutable** | Yes | No | Yes | Yes |
| **Duplicates** | Allowed | Allowed | Not allowed | Keys: No, Values: Yes |
| **Indexing** | Yes | Yes | No | By key |
| **Stores** | Values | Values | Unique values | Key-value pairs |
| **Use when** | You need a changeable sequence | You need an unchangeable sequence | You need unique items | You need key-value data |

### Examples

```python
# List
my_list = [10, 20, 20, 30]

# Tuple
my_tuple = (10, 20, 20, 30)

# Set
my_set = {10, 20, 30}

# Dictionary
my_dict = {"name": "John", "age": 25}
